<h3>Import requirements</h3>

```python
from textblob import TextBlob
from textblob import Word
from textblob.wordnet import VERB
```

```python
wiki.noun_phrases
```

<h2 id="sec3">3. Sentiment analysis</h2>

<p>Sentiment of a sentence is defined as its polarity and subjectivity, with ranges between [-1,1] and [0,1] respectively.</p>

```python
testimonial = TextBlob("Textblob is amazingly simple to use. What great fun!")
```

```python
testimonial.sentiment
```

```python
testimonial.sentiment.polarity
```

<h2 id="sec4">4. Tokenization</h2>

```python
zen = TextBlob(
    "Beautiful is better than ugly. "
    "Explicit is better than implicit. "
    "Simple is better than complex."
)
```

<p>Use .words and .sentences to break down a text blob</p>

```python
zen.words
```

```python
zen.sentences
```

```python
for sentence in zen.sentences:
    print(sentence.sentiment)
```

<h2 id="sec5">5. Words inflection and lemmatization</h2>

<p>Each word in a text blob is a word object which has its own methods such as .singularize() and .pluralize()

```python
sentence = TextBlob("Use 4 spaces per indentation level.")
sentence.words
```

```python
sentence.words[2].singularize()
```

```python
sentence.words[-1].pluralize()
```

<p>Lemmatize means to reduce a word back to its base form, for example ran and runs turn into to run</p>

```python
w = Word("octopi")
w.lemmatize()
```

```python
w = Word("went")
w.lemmatize("v")
```

<h2 id="sec6">6. WordNet</h2>

```python
Word("hack").get_synsets(pos=VERB)
```

```python
Word("octopus").definitions
```

<h2 id="sec8">8. Spelling correction</h2>

<p>The correct() method can be used to try and correct spelling errors</p>

```python
b = TextBlob("I havv goood speling!")
print(b.correct())
```

<p>You can also spellcheck to get a listof suggestion tuples with (word, confidence)</p>

```python
w = Word("falibility")
w.spellcheck()
```

```python
w = Word("askd")
w.spellcheck()
```

<h2 id="sec9">9. Word and phrase frequencies</h2>

<p>There are multiple methods to get the frequencies</p>

```python
monty = TextBlob("We are no longer the Knights who say Ni. "
                    "We are now the Knights who say Ekki ekki ekki PTANG.")
```

```python
monty.word_counts['ekki']
```

```python
monty.words.count('ekki')
```

```python
monty.words.count('ekki', case_sensitive=True)
```

<p>And we can also use it with noun phrases like before</p>

```python
wiki.noun_phrases.count('python')
```

<h2 id="sec10">10. Parsing text blobs</h2>

<h2 id="sec11">11. Text blobs like python strings</h2>

<p>We can use common string indexing and methods on text blobs</p>

```python
zen[0:19]
```

```python
zen.upper()
```

```python
zen.find("Simple")
```

<p>We can make string comparisons</p>

```python
apple_blob = TextBlob("apples")
banana_blob = TextBlob("bananas")
apple_blob < banana_blob
```

```python
apple_blob == "apples"
```

<p>We can concatenate and format strings</p>

```python
apple_blob + " and " + banana_blob
```

```python
"{0} and {1}".format(apple_blob, banana_blob)
```

<p>Something new we can do is get the start and end indices of sentences</p>

```python
for s in zen.sentences:
    print(s)
    print("---- Starts at index {}, Ends at index {}".format(s.start, s.end))
```

<h2 id="sec12">12. n-grams</h2>

<p>n-grams allow us to get all chunks of n-words from a sentence</p>

```python
blob = TextBlob("Now is better than never.")
blob.ngrams(n=3)
```
